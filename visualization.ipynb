```python
import numpy as np
```

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("titanic")
df
```

```python
pclass_mean_fare = df.groupby('pclass')['fare'].mean()
```

```python
plt.figure(figsize=(6,4), dpi=100)
plt.plot(pclass_mean_fare.index, pclass_mean_fare.values, marker='o', color='navy')
plt.title('Fare by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(6, 4), dpi=100)
sns.lineplot(x ='pclass', y='fare', data=df, estimator="mean", marker='o', color='navy')
plt.title('Pclass by Fare')
plt.xlabel('Fare')
plt.ylabel('Pclass')
plt.show()
```

```python
counts = df['embark_town'].value_counts()
counts
```

```python
plt.figure(figsize=(6,4))
plt.bar(counts.index, counts.values)
plt.title('number of passengers by embark town')
plt.xlabel('embark town')
plt.ylabel("number of passengers")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='embark_town', data=df)
plt.title('number of passengers by embark town')
plt.xlabel('embark town')
plt.ylabel("number of passengers")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='embark_town', hue='alive')
plt.show()
```

```python
counts = df["embark_town"].value_counts()
counts
```

```python
plt.figure(figsize=(6, 6))
plt.pie(counts.values, labels=counts.index, autopct='%.2f%%', startangle=0)
plt.title('pie chart of passenger proportion')
plt.legend(loc="lower left")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df['age'], bins=40, edgecolor='black')
plt.title('Age distribution')
plt.xlabel('Age')
plt.ylabel('Counts')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=df, x="age", bins=20, hue='alive', multiple='dodge')
plt.title('Age distribution of the victim')
plt.xlabel('Age')
plt.ylabel('Counts')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='pclass', y='age')
plt.title('Age by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Age')
plt.show()
```
