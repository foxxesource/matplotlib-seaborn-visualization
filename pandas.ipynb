```python
df = sns.load_dataset("tips")
```

```python
df
```

```python
grouped = df.groupby('day', observed=False).agg(total_customers=('smoker', 'count'),smoker_count=('smoker', lambda x: (x == 'Yes').sum()))
```

```python
grouped['smoker_proportion'] = grouped['smoker_count'] / grouped['total_customers']
```

```python
sorted_days = grouped.sort_values('smoker_proportion', ascending=False)
```

```python
result = sorted_days[['smoker_proportion']]
print(result)
```
